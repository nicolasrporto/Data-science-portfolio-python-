```python
import numpy as np
import matplotlib.pyplot as plt



N = int(input()) # Numero de agulhas ou sorteios
seed = int(input()) # semente dos números aleatórios


# Os vetores abaixo devem receber 'append' dos valores estimados a cada potência de 10, começando do próprio 10.
pi1 = [] # vetor que deve guardar os valores de pi estimados pelo primeiro método
pi2 = [] # vetor que deve guardar os valores de pi estimados pelo segundo método

# Inicialização do gerador de números pseudoaleatórios
np.random.seed(seed)

Ncirc = 0 #contagem de agulhas no círculo
soma_fx = 0 #contagem valor_medio
potencias_N = []

for i in range(1,N+1):



    x = np.random.uniform(0.0, 1.0)
    y = np.random.uniform(0.0, 1.0)

    #agulhas de buffon

    if (y<=np.sqrt(1-x**2)):
        Ncirc+=1

    soma_fx += np.sqrt(1-x**2)

    if (np.mod(np.log10(i),1) == 0):

      pi1_valor = 4*(Ncirc/i)
      pi1.append(pi1_valor)


      #valormédio
      pi2_valor = 4.0 * soma_fx / (i)
      pi2.append(pi2_valor)

      potencias_N.append(i)


print("Metodo 1:")
for j in range(len(pi1)):
    print("{:.8f}".format(pi1[j]))
print("Metodo 2:")
for j in range(len(pi2)):
    print("{:.8f}".format(pi2[j]))

    plt.figure(figsize=(9, 5))

# Plot das estimativas
plt.plot(potencias_N, pi1, marker='o', linestyle='-', color='#1f77b4', label='Método 1 (Acerto/Erro)')
plt.plot(potencias_N, pi2, marker='s', linestyle='--', color='#ff7f0e', label='Método 2 (Valor Médio)')

# Linha horizontal indicando o valor real de pi
plt.axhline(y=np.pi, color='#8c1d1d', linestyle=':', label='Valor Real ($\pi$)')

# Escala logarítmica no eixo X para acomodar as potências de 10
plt.xscale('log')

# Configurações de títulos e eixos
plt.title('Evolução das Estimativas de $\pi$')
plt.xlabel('Número de Sorteios ($N$)')
plt.ylabel('Valor Estimado')

plt.xticks(potencias_N, labels=[str(p) for p in potencias_N])

# Ativação das grades principal e secundária (escala logarítmica)
plt.grid(True, which="both", linestyle='-', alpha=0.2)


plt.legend()
plt.tight_layout()

plt.show()
```
