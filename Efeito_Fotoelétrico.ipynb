**Medidas Rápidas - com ruido**

```python
from google.colab import files
import pandas as pd
import io
import matplotlib as plt

#Leitura dos arquivos
df_amarelo = pd.read_csv('Amarelo (rap).csv', delimiter=';', encoding='latin-1', decimal='.')
df_azul = pd.read_csv('Azul (rap).csv', delimiter=';', encoding='latin-1', decimal='.')
df_ultravioleta = pd.read_csv('Ultravioleta (rap).csv', delimiter=';', encoding='latin-1', decimal='.')
df_verde = pd.read_csv('Verde (rap).csv', delimiter=';', encoding='latin-1', decimal='.')
df_violeta = pd.read_csv('Violeta (rap).csv', delimiter=';', encoding='latin-1', decimal='.')
df_ruido = pd.read_csv('Ruido (rap).csv', delimiter=';', encoding='latin-1', decimal='.')
```

```python
# Substituir vírgulas por pontos em todas as colunas numéricas
df_amarelo.replace(to_replace={',': '.'}, regex=True, inplace=True)
df_azul.replace(to_replace={',': '.'}, regex=True, inplace=True)
df_ultravioleta.replace(to_replace={',': '.'}, regex=True, inplace=True)
df_verde.replace(to_replace={',': '.'}, regex=True, inplace=True)
df_violeta.replace(to_replace={',': '.'}, regex=True, inplace=True)
df_ruido.replace(to_replace={',': '.'}, regex=True, inplace=True)
df_ruido.head(10)
```

```python
import pandas as pd
import matplotlib.pyplot as plt


# Lista de DataFrames
dataframes = [df_amarelo, df_azul, df_ultravioleta, df_verde, df_violeta]

# Cores para cada DataFrame
cores = ['yellow', 'blue', 'purple', 'green', 'violet']

# Configurações para o gráfico
fig, ax = plt.subplots()

# Iterar sobre os DataFrames e plotar a coluna 1 versus coluna 2
for i, df in enumerate(dataframes):
    try:
        df[df.columns[:2]] = df[df.columns[:2]].replace(to_replace={',': '.'}, regex=True).astype(float)
        df.plot(x=df.columns[0], y=df.columns[1], ax=ax, label=cores[i], color=cores[i])
    except KeyError as e:
        print(f'Erro: {e} não encontrado. Verifique os nomes das colunas no DataFrame {i+1}.')
    except ValueError:
        print(f'Erro: DataFrame {i+1} contém valores não numéricos em {df.columns[0]} ou {df.columns[1]}.')

# Configurações do gráfico
plt.title('Relação entre Tensão e corrente para diferentes comprimentos de onda')
plt.xlabel('Tensão [V]')
plt.ylabel('Corrente [A]')
# Removendo o e-8 do eixo y
ax.yaxis.set_major_formatter(ScalarFormatter(useMathText=True))
ax.ticklabel_format(axis='y', style='sci', scilimits=(-2,2))

plt.legend(['Amarelo', 'Azul', 'Ultravioleta', 'Verde', 'Violeta'])
plt.show()
```

```python
 import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
####AMARELo
# Escolha as colunas que representam os pontos x e y
x_column = df_amarelo_.columns[11]
y_column = df_amarelo_.columns[12]

# Defina o intervalo de pontos de x do DataFrame
intervalo_inicio = 98
intervalo_fim = 107

# Selecionar os pontos dentro do intervalo
intervalo_x = df_amarelo_[x_column][intervalo_inicio:intervalo_fim+1]
intervalo_y = df_amarelo_[y_column][intervalo_inicio:intervalo_fim+1]

# Calcular a linha reta ajustada usando o método dos mínimos quadrados
A = np.vstack([intervalo_x, np.ones(len(intervalo_x))]).T
m, c = np.linalg.lstsq(A, intervalo_y, rcond=None)[0]

# Calcular o ponto onde a reta atravessa y=0
x_intercept = -c / m

# Estender a reta para os dois lados
x_prolongado = np.array([x_intercept - 5, x_intercept + 5])
y_prolongado = m * x_prolongado + c

# Plotar o gráfico
plt.figure(figsize=(10, 6))
plt.plot(df_amarelo_[x_column], df_amarelo_[y_column], marker='o', markersize=3, linestyle='', label='Pontos')
plt.plot(x_prolongado, y_prolongado, color='black', linestyle='--', linewidth=1, label='Reta ajustada e prolongada')
plt.plot(intervalo_x, intervalo_y, marker='o', markersize=5, linestyle='', color='red', label='Intervalo selecionado')
plt.xlabel('Tensão [A]')
plt.ylabel('Corrente [A]')
plt.title('Amarelo')
plt.axhline(0, color='black', linewidth=0.5)
plt.legend()
plt.grid(True)
plt.xlim(-1, 0.8)  # Definindo os limites do eixo x
plt.ylim(-5e-11,1.2e-8)  # Definindo os limites do eixo x
plt.show()
```
